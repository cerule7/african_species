```python
import pandas as pd
from ftfy import fix_text, fix_and_explain

DESCRIPTION_CSV = "wfo_descriptions.csv"
OUTPUT_CSV = "cleaned_labeled_descriptions.csv"

description_df = pd.read_csv(DESCRIPTION_CSV)

valid_types = ["Morphology", "General Information", "Diagnostic Description", "Habit"]

filtered = description_df[
    description_df["description_header"].isin(valid_types) &
    description_df["description"].notna()
].copy()

print(filtered['description_header'].value_counts())
```

```python
for row in filtered['description']:
    explanation = fix_and_explain(row, uncurl_quotes=False)
    if len(explanation.explanation) > 0:
        print(f"Original: {row}")
        print(f"Fixed: {fix_text(row)}")
        print(f"Explanation: {explanation.explanation}")
```

```python
filtered["cleaned_description"] = filtered["description"].apply(fix_text)

# No longer needed
filtered.drop('description', axis=1)
```

```python
from lingua import LanguageDetectorBuilder, Language

languages = [Language.ENGLISH, Language.FRENCH, Language.GERMAN, Language.SPANISH, Language.PORTUGUESE, Language.AFRIKAANS, Language.LATIN]

detector = LanguageDetectorBuilder.from_languages(*languages).build()

for idx, row in filtered['cleaned_description'].items():
    confidence_values = detector.compute_language_confidence_values(row)

    if confidence_values[0].value >= 0.75:
        filtered.loc[idx, 'detected_language'] = (
            confidence_values[0].language.name
        )
    else:
        filtered.loc[idx, 'detected_language'] = 'NOT_DETECTED'
```

```python
summary = pd.DataFrame({
    "count": filtered["detected_language"].value_counts(),
    "percent": filtered["detected_language"].value_counts(normalize=True) * 100
})

print(summary)
```

```python
# The 'NOT_DETECTED' descriptions are English so shall be treated as such
print(filtered[filtered["detected_language"] == "NOT_DETECTED"]["cleaned_description"])
```
