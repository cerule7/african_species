```python
import pandas as pd
from ftfy import fix_text, fix_and_explain

DESCRIPTION_CSV = "wfo_descriptions.csv"

description_df = pd.read_csv(DESCRIPTION_CSV)

valid_types = ["Morphology", "General Information", "Diagnostic Description", "Habit"]

filtered = description_df[
    description_df["description_header"].isin(valid_types) &
    description_df["description"].notna()
].copy()
```

```python
for row in filtered['description']:
    explanation = fix_and_explain(row, uncurl_quotes=False)
    if len(explanation.explanation) > 0:
        print(f"Original: {row}")
        print(f"Fixed: {fix_text(row)}")
        print(f"Explanation: {explanation.explanation}")
```
